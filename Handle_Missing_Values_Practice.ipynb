## Objective
Analyze missing values, clean the dataset, and explain the chosen treatment.

```python
import pandas as pd

df = pd.read_csv('titanic_missing_values.csv')
df.head()
```

## 1. Original missing-value analysis

```python
print('Dataset shape:', df.shape)
print(df.isnull().sum())
print('Total missing values:', df.isnull().sum().sum())
```

## 2. Missing-value percentage

```python
print((df.isnull().mean() * 100).round(2))
```

## 3. Fill numerical columns

```python
# Median is less sensitive to extreme values than the mean.
cleaned_df = df.copy()
age_median = cleaned_df['age'].median()
fare_median = cleaned_df['fare'].median()
cleaned_df['age'] = cleaned_df['age'].fillna(age_median)
cleaned_df['fare'] = cleaned_df['fare'].fillna(fare_median)
print('Age median:', age_median, '| Fare median:', fare_median)
```

## 4. Fill categorical column

```python
# Mode is the most frequent category, suitable for this categorical column.
embarked_mode = cleaned_df['embarked'].mode()[0]
cleaned_df['embarked'] = cleaned_df['embarked'].fillna(embarked_mode)
print('Embarked mode:', embarked_mode)
```

## 5. Compare before and after

```python
print(pd.DataFrame({'before': df.isnull().sum(), 'after': cleaned_df.isnull().sum()}))
print('Remaining missing values:', cleaned_df.isnull().sum().sum())
```

## 6. Demonstrate dropna()

```python
dropped_df = df.dropna()
print('Original rows:', len(df))
print('Rows after dropna():', len(dropped_df))
```

## 7. Save cleaned dataset

```python
cleaned_df.to_csv('titanic_cleaned.csv', index=False)
print('Saved titanic_cleaned.csv')
```

## Key Observations
1. Missing values occur in age, fare, and embarked.
2. Median imputation is used for numerical columns because it is less affected by extreme values.
3. Mode imputation is used for the categorical embarked column.
4. `fillna()` retains rows while replacing missing values.
5. `dropna()` can remove useful records, so it should be used thoughtfully.
6. Missing-value counts are compared before and after cleaning.

## Conclusion
Choosing treatment based on column type and distribution helps prepare a dataset for analysis and machine learning.
